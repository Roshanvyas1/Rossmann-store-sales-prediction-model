## Preprocessing and Feature Engineering

```python
train_df.info()
```

```python
train_df['StateHoliday'] = train_df['StateHoliday'].astype('str')
```

### Date

```python
def split_date(df):
    df['Date'] = pd.to_datetime(df['Date'])
    df['Year'] = df.Date.dt.year
    df['Month'] = df.Date.dt.month
    df['Day'] = df.Date.dt.day
    df['WeekOfYear'] = df.Date.dt.isocalendar().week
```

```python
split_date(train_df)
split_date(test_df)
```

### Open/Closed Stores

```python
train_df[train_df.Open == 0].Sales.value_counts()
```

```python
train_df = train_df[train_df.Open == 1]
train_df
```

### Competition months since open

```python
def comp_months(df):
    df['CompetitionOpenMonths'] = 12 * (df.Year - df.CompetitionOpenSinceYear) + (df.Month - df.CompetitionOpenSinceMonth)
    df['CompetitionOpenMonths'] = df['CompetitionOpenMonths'].map(lambda x: 0 if x < 0 else x).fillna(0)
```

```python
comp_months(train_df)
comp_months(test_df)
train_df.sample(10)
```

### Promo2 Weeks since Open and Promo2 Month or not

```python
def check_promo_month(row):
    month2str = {1:'Jan', 2:'Feb', 3:'Mar', 4:'Apr', 5:'May', 6:'Jun',              
                 7:'Jul', 8:'Aug', 9:'Sept', 10:'Oct', 11:'Nov', 12:'Dec'}
    try:
        months = (row['PromoInterval'] or '').split(',')
        if row['Promo2OpenWeeks'] and month2str[row['Month']] in months:
            return 1
        else:
            return 0
    except Exception:
        return 0

def promo_cols(df):
    # Months since Promo2 was open
    df['Promo2OpenWeeks'] = 12 * (df.Year - df.Promo2SinceYear) +  (df.WeekOfYear - df.Promo2SinceWeek)*7/30.5
    df['Promo2OpenWeeks'] = df['Promo2OpenWeeks'].map(lambda x: 0 if x < 0 else x).fillna(0) * df['Promo2']
    # Whether a new round of promotions was started in the current month
    df['IsPromo2Month'] = df.apply(check_promo_month, axis=1) * df['Promo2']
```

```python
promo_cols(train_df)
promo_cols(test_df)
train_df.sample(10)[['Date', 'Promo2', 'PromoInterval', 'IsPromo2Month']]
```

## Inputs and Target Columns

```python
train_df.columns
```

```python
input_cols = ['Store', 'DayOfWeek', 'Promo',
       'StateHoliday', 'SchoolHoliday', 'StoreType', 'Assortment',
       'CompetitionDistance', 'Promo2', 'Year', 'Month', 'Day',
       'WeekOfYear', 'CompetitionOpenMonths', 'Promo2OpenWeeks',
       'IsPromo2Month']
target_col = 'Sales'
```

```python
categorical_cols = ['Store', 'DayOfWeek', 'StateHoliday', 'StoreType', 'Assortment']

# categories learned from train only
cat_types = {c: pd.CategoricalDtype(sorted(train_df[c].unique())) for c in categorical_cols}

for df in (train_df, test_df):
    for c in categorical_cols:
        df[c] = df[c].astype(cat_types[c])

train_df.dtypes
```

```python
train_df = train_df.sort_values(by='Date', ignore_index=True)
train_df
```

```python
inputs = train_df[input_cols]
test_inputs = test_df[input_cols]
inputs.head()
```

```python
numeric_cols = ['CompetitionDistance', 'CompetitionOpenMonths', 'Promo2OpenWeeks',
              'Day', 'Month', 'Year', 'WeekOfYear']
binary_cols = ['Promo', 'SchoolHoliday', 'IsPromo2Month', 'Promo2']
categorical_cols = ['Store', 'DayOfWeek', 'StateHoliday', 'StoreType', 'Assortment']
```

## KFold Validation

```python
from sklearn.model_selection import KFold
```

```python
def train_and_evaluate(X_train, Y_train, X_val, Y_val, **params):
    model = XGBRegressor(random_state=42, n_jobs=-1, **params)
    model.fit(X_train, Y_train)
    train_rmse = rmse(model.predict(X_train), Y_train)
    val_rmse = rmse(model.predict(X_val), Y_val)
    return model, train_rmse, val_rmse
```

```python
kfold = KFold(n_splits=5)
```

```python
models = []

for train_idxs, val_idxs in kfold.split(X):
    X_train, train_targets = X.iloc[train_idxs], Y.iloc[train_idxs]
    X_val, val_targets = X.iloc[val_idxs], Y.iloc[val_idxs]
    model, train_rmse, val_rmse = train_and_evaluate(X_train, 
                                                     train_targets, 
                                                     X_val, 
                                                     val_targets, 
                                                     max_depth=4, 
                                                     n_estimators=20)
    models.append(model)
    print(f"Train RMSE: {train_rmse}, Val RMSE: {val_rmse}")
```

### Avg predictions from 5 different models.

```python
def predict_avg(models, inputs):
    return np.mean([model.predict(inputs) for model in models], axis=0)
```

```python
X_preds = predict_avg(models, X)
```

```python
X_preds
```

```python
rmse(X_preds, Y)
```

## Hyperparameter Tuning for xgboost

### Spliting train and val dataset time ordered

```python
size = int(len(X) * 0.75)
X_train, X_val = X[:size], X[size:]
Y_train, Y_val = Y[:size], Y[size:]
```

```python
def test_params(**params):
    model = XGBRegressor(n_estimators=1000, 
                         early_stopping_rounds=20, 
                         eval_metric='rmse',
                         n_jobs=-1, 
                         random_state=42, 
                         verbosity=2, 
                         **params)
    model.fit(X_train, Y_train, eval_set=[(X_val, Y_val)], verbose=50)
    train_rmse = rmse(model.predict(X_train), Y_train)
    val_rmse = rmse(model.predict(X_val), Y_val)
    print('Train RMSE: {}, Validation RMSE: {}'.format(train_rmse, val_rmse))
```

### learning_rate

```python
test_params(learning_rate=0.1) # Performed best
test_params(learning_rate=0.3)
test_params(learning_rate=0.5)
test_params(learning_rate=0.9)
test_params(learning_rate=0.99)
```

### min_child_weight

```python
test_params(min_child_weight=1)
test_params(min_child_weight=2)
test_params(min_child_weight=3) 
test_params(min_child_weight=5) # Perfomed best
```

### subsample

```python
test_params(subsample=0.5)
test_params(subsample=0.7) # Performed best
test_params(subsample=0.9)
test_params(subsample=1) 
```

### colsample_bytree

```python
test_params(colsample_bytree=0.5)
test_params(colsample_bytree=0.7)
test_params(colsample_bytree=0.9) # Performed best
test_params(colsample_bytree=1)
```

### max_depth

```python
test_params(max_depth=3)
test_params(max_depth=4) # Performed best
test_params(max_depth=5)
test_params(max_depth=8)
```

## Final Hypertuned model

```python
model = XGBRegressor(
    n_estimators=2000, 
    early_stopping_rounds=50,
    learning_rate=0.2,
    min_child_weight=2,
    subsample=0.8,
    colsample_bytree=0.9,
    eval_metric='rmse', 
    max_depth=4,
    n_jobs=-1, 
    random_state=42
)
model.fit(X_train, Y_train, eval_set=[(X_val, Y_val)], verbose=50)
train_rmse = rmse(model.predict(X_train), Y_train)
val_rmse = rmse(model.predict(X_val), Y_val)
print('Train RMSE: {}, Validation RMSE: {}'.format(train_rmse, val_rmse))
```
